# 02 — Baseline LDA

Fit Latent Dirichlet Allocation (gensim) on arXiv abstracts and score **C_v / NPMI / UMass** coherence. This is the baseline the professor asked to beat with semantic embeddings.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

from src.config import DEFAULT_N_DOCS, RANDOM_SEED
from src.data import load_corpus
from src.preprocess import preprocess_frame
from src.lda_baseline import fit_lda
from src.evaluate import topic_coherence, topic_diversity, summarize_topics
```

```python
df = preprocess_frame(load_corpus("arxiv", n_docs=DEFAULT_N_DOCS, seed=RANDOM_SEED))
texts = df["tokens"].tolist()
lda = fit_lda(texts, n_topics=12, seed=RANDOM_SEED)
coh = topic_coherence(lda.topic_words, texts, dictionary=lda.dictionary)
print("Coherence:", coh)
print("Diversity:", topic_diversity(lda.topic_words))
summarize_topics(lda.topic_words)
```

LDA models each document as a mixture of topics (`θ`) and each topic as a distribution over words (`φ`). Coherence `C_v` measures how often the top words of a topic co-occur in a sliding window — higher is more interpretable.
